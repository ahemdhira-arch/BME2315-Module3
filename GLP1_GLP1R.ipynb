```python

import numpy as np
import matplotlib.pyplot as plt

def Fact(GLP1, sema):
    return (GLP1/0.17 + sema/0.38) / (1 + GLP1/0.17 + sema/0.38)

sema = np.logspace(-2, 2, 500)

fasting = Fact(0.015, sema)
fed = Fact(0.040, sema)

plt.figure()
plt.semilogx(sema, fasting, label="Fasting GLP1")
plt.semilogx(sema, fed, label="Fed GLP1")

plt.xlabel("Semaglutide concentration (nM)")
plt.ylabel("Fraction active (Fact)")
plt.title("GLP1R Activation")
plt.ylim(0, 1)
plt.legend()
plt.show()
```


<h2 style="color:#1f77b4;">GLP1R Activation Analysis</h2>

<h3 style="color:#ff7f0e;">Interpretation</h3>

The graph shows that GLP1R activation is higher at fed GLP1 levels (19.0%) compared to fasting levels (8.1%) without semaglutide, but as semaglutide concentration increases, both responses get closer to 100%.
